```python
import numpy as np
house_size = np.array([600,750,850,1000,1100,1250,1400,1550,1700,1850])
bedroom= np.array([1,2,2,2,3,3,3,4,4,4])
price= np.array([28,34,38,44,49,55,60,68,73,79])
```

```python
x= np.column_stack((np.ones(10),house_size,bedroom))
print(x)
```

```python
X_transpose_X= x.T @ x
print(X_transpose_X)
```

```python
inverse=np.linalg.inv(X_transpose_X)
print(inverse)
```

```python
X_transpose_Y=x.T @ price
print(X_transpose_Y)
```

```python
beta =inverse @ X_transpose_Y
print(beta)
```

```python
new_house=np.array([1,1300,3])
```

```python
prediction =new_house @ beta 
print(prediction)
```

## Conclusion

Using the Normal Equation, I trained a linear regression model using house size and number of bedrooms.

For a 1300 sq. ft. house with 3 bedrooms, the model predicted a price of approximately 56.68 lakhs.

```python
import numpy as np
new_house = np.array([1, 1300, 3])

prediction = new_house @ beta

print("Predicted price:", prediction)
```
