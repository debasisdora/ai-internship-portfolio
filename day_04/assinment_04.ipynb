```python
import pandas as pd
df=pd.read_csv("student_messy.csv")
```

```python
df["department"] = df["department"].str.strip()
df["department"] = df["department"].str.title()
```

```python
dept_avg = df.groupby("department")["marks"].mean().reset_index(name="avg_marks")
dept_avg
```

```python
merged = df.merge(dept_avg, on="department")
merged
```

```python
result = merged[
    ((merged["department"] == "Computer Science") |
     (merged["department"] == "Mechanical")) &
    (merged["marks"] < merged["avg_marks"])
]

result
```
