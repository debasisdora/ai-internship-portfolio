```python
from io import StringIO

data = """order_id,customer,city,amount,status
5001,Rahul Verma,Mumbai,2500,Delivered
5002,Fatima Sheikh ,mumbai,1800,Delivered
5003,Suresh Iyer,Bangalore,,Delivered
5004,Meena Joshi,BANGALORE,3200,Cancelled
5001,Rahul Verma,Mumbai,2500,Delivered
5006,Aakash Bose,Pune,4100,Delivered
5007,Divya Nair,Chennai,,Pending
"""

df = pd.read_csv(StringIO(data))
df
```

```python
df.duplicated()
```

```python
df =df.drop_duplicates()
df
```

```python
df['customer']=df['customer'].str.strip()
df['city']=df['city'].str.strip()
df
```

```python
df['city']=df['city'].str.title()
df
```

```python
city_avg=df.groupby('city')['amount'].transform('mean')
df['amount']=df['amount'].fillna(city_avg)
df
```
