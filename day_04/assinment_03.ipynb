```python
import pandas as pd
df=pd.read_csv("student_messy.csv")
departments=pd.read_csv("departments.csv")
departments
```

```python
df["department"] = df["department"].str.strip()
df["department"] = df["department"].str.title()
```

```python
left_merge = df.merge(departments, on="department", how="left")
left_merge
```

```python
inner_merge = df.merge(departments, on="department", how="inner")
inner_merge
```

```python
print("Left Merge Rows:", len(left_merge))
print("Inner Merge Rows:", len(inner_merge))
```
